## 3. Git y GitHub

**Git** es un programa que guarda fotos de tu proyecto a lo largo del tiempo. **GitHub** es un sitio web que guarda una copia de esas fotos en línea. El vocabulario mínimo:

| Término | Qué es |
|---|---|
| repositorio (*repository*) | Una carpeta cuyo historial controla Git |
| *commit* | Una foto del repositorio, con autor, fecha y un mensaje que dice qué cambió |
| historial (*log*) | La lista de *commits*, del más nuevo al más viejo |
| diferencia (*diff*) | Qué líneas cambiaron entre dos *commits* |
| remoto (*remote*) | La copia del repositorio en otro lugar; para nosotros, GitHub |

Por qué nos importa:

- **Tu entrega es un *commit***: lo que esté en tu repositorio al inicio de la clase.
- **No pierdes trabajo**: cualquier versión anterior sigue en el historial.
- **El historial muestra cómo trabajaste**: muchos *commits* pequeños con buenos mensajes valen más que uno enorme la noche anterior.

### 3.1 Git en vivo

Vamos a usar Git de verdad, en una carpeta temporal. La función `git` ejecuta un comando de Git dentro de esa carpeta y muestra lo que responde. No necesitas aprender estos comandos hoy: Colab los hará por ti. Mira qué pasa en cada paso.

```python
repo = Path(tempfile.mkdtemp()) / "mi_proyecto"
repo.mkdir()


def git(*args):
    """Ejecuta `git <args>` dentro de `repo` e imprime la respuesta."""
    identity = ["-c", "user.name=Estudiante", "-c", "user.email=estudiante@example.com"]
    result = subprocess.run(["git", *identity, *args], cwd=repo, capture_output=True, text=True)
    output = (result.stdout + result.stderr).strip()
    if output:
        print(output)


git("init", "-b", "main")
```

Creamos un archivo. Git lo ve, pero todavía no lo guarda: aparece como *untracked* (sin seguimiento).

```python
(repo / "notas.md").write_text("# Mis notas\n\nSemana 1: Git guarda fotos del proyecto.\n")
git("status", "--short")
```

Guardar una foto tiene dos pasos: `add` elige qué archivos entran en la foto y `commit` la toma, con un mensaje.

```python
git("add", "notas.md")
git("commit", "-m", "Agregar notas de la semana 1")
git("log", "--oneline")
```

Cambiamos el archivo. `diff` muestra qué líneas cambiaron respecto de la última foto: `+` las agregadas, `-` las borradas.

```python
(repo / "notas.md").write_text(
    "# Mis notas\n\nSemana 1: Git guarda fotos del proyecto.\nSemana 2: NumPy y álgebra lineal.\n"
)
git("diff")
```

```python
git("commit", "-am", "Agregar la semana 2")
git("log", "--oneline")
```

## 4. Python orientado a objetos

### 4.1 Por qué lo necesitamos

En la *programación orientada a objetos* (*object-oriented programming*, POO) agrupamos **datos** y las **funciones que trabajan con ellos** en un mismo objeto. Ya usaste objetos aunque no lo supieras. Un modelo de sklearn es uno:

```python
X = rng.normal(size=(50, 1))
y = 3 * X[:, 0] + 1 + 0.1 * rng.normal(size=50)

model = LinearRegression()  # crear el objeto
model.fit(X, y)             # un método: modifica el estado del objeto
print("pendiente:", model.coef_.round(2), " ordenada:", model.intercept_.round(2))  # atributos: el estado
print("predicción para x = 2:", model.predict([[2.0]]).round(2))
```

`model` guarda **estado** (los coeficientes aprendidos, en atributos como `coef_`) y tiene **métodos** (`fit`, `predict`) que lo usan o lo cambian. En PyTorch, desde la semana 8, **cada modelo, cada capa y cada conjunto de datos es una clase que escribes tú**. Hoy aprendemos a escribirlas.

### 4.2 Una clase mínima

Una **clase** (*class*) es el molde; cada objeto creado con ella es una **instancia** (*instance*). Al crear una instancia, Python llama a `__init__`, que guarda los datos iniciales en `self`. `self` es la instancia misma: es lo que permite que cada objeto tenga sus propios datos.

```python
class TrainingLog:
    """Guarda la pérdida de cada época de un entrenamiento."""

    def __init__(self, name):
        self.name = name   # atributo: un dato que pertenece a la instancia
        self.losses = []

    def add(self, loss):   # método: una función que recibe la instancia como `self`
        self.losses.append(loss)

    def best(self):
        return min(self.losses)


log_a = TrainingLog("modelo A")
log_b = TrainingLog("modelo B")
for loss in [0.9, 0.5, 0.4, 0.45]:
    log_a.add(loss)
log_b.add(1.2)

print(log_a.name, log_a.losses, "→ mejor:", log_a.best())
print(log_b.name, log_b.losses)  # cada instancia tiene sus propios datos
```

### 4.3 Métodos especiales

Los métodos con doble guion bajo (*dunder methods*, de *double underscore*) conectan tu clase con la sintaxis de Python: `__len__` hace funcionar `len(obj)`, `__repr__` decide cómo se imprime el objeto, `__getitem__` hace funcionar `obj[i]` y `__call__` permite llamar al objeto como si fuera una función. **Herencia** (*inheritance*): `class B(A)` crea una clase `B` que recibe todo lo de `A` y puede agregar o reemplazar métodos.

```python
class PrettyTrainingLog(TrainingLog):  # hereda __init__, add y best
    def __len__(self):
        return len(self.losses)

    def __repr__(self):
        return f"PrettyTrainingLog({self.name!r}, {len(self)} épocas, mejor={self.best():.2f})"


log = PrettyTrainingLog("modelo C")
for loss in [0.8, 0.6, 0.7]:
    log.add(loss)
print(len(log))
log
```

### 4.4 Una capa lineal

Una capa lineal transforma cada ejemplo $x$ en $Wx + b$. Con un lote de ejemplos en las filas de una matriz `x` de forma `(lote, entradas)`, la capa entera es una sola multiplicación de matrices:

$$y = x W^\top + b, \qquad W: (\text{salidas}, \text{entradas}), \quad b: (\text{salidas},)$$

Usamos la misma convención de formas que PyTorch. De NumPy hoy solo necesitas `@` (producto de matrices), `.T` (transpuesta) y `.shape` (la forma); la semana 2 lo vemos a fondo.

Empezamos por una clase base `Module` que define lo que toda capa sabe hacer:

- `__call__` llama a `forward`, para escribir `layer(x)` en lugar de `layer.forward(x)`. PyTorch hace lo mismo.
- `forward` lanza un error: cada subclase debe reemplazarlo con su cálculo.
- `parameters` devuelve la lista de arreglos que el entrenamiento ajustará. Por defecto, ninguno.
- `__repr__` imprime el nombre de la clase; `Linear` lo reemplaza para mostrar también sus formas.

```python
class Module:
    def __call__(self, x):
        return self.forward(x)

    def forward(self, x):
        raise NotImplementedError(f"{type(self).__name__} debe definir forward")

    def parameters(self):
        return []

    def __repr__(self):
        return f"{type(self).__name__}()"


class Linear(Module):
    def __init__(self, in_features, out_features, rng):
        # Pesos aleatorios pequeños; en la semana 9 veremos por qué se escalan así.
        self.weight = rng.normal(size=(out_features, in_features)) / np.sqrt(in_features)
        self.bias = np.zeros(out_features)

    def forward(self, x):
        return x @ self.weight.T + self.bias

    def parameters(self):
        return [self.weight, self.bias]

    def __repr__(self):
        out_features, in_features = self.weight.shape
        return f"Linear({in_features} → {out_features})"


layer = Linear(3, 2, rng)
x = rng.normal(size=(4, 3))  # un lote de 4 ejemplos con 3 variables cada uno
out = layer(x)               # __call__ → forward
print(layer, "| x:", x.shape, "→ salida:", out.shape)
print("formas de los parámetros:", [p.shape for p in layer.parameters()])
```

Ahora una capa sin parámetros, la activación **ReLU** (*rectified linear unit*): deja pasar los positivos y cambia los negativos por cero. Como hereda de `Module`, ya sabe responder a `__call__` y a `parameters()`.

```python
class ReLU(Module):
    def forward(self, x):
        return np.maximum(x, 0)


relu = ReLU()
print(relu(np.array([-2.0, -0.5, 0.0, 1.5])))
print("parámetros:", relu.parameters())
```

### 4.5 Composición: una red es una lista de capas

`Sequential` aplica sus capas en orden y junta los parámetros de todas. Es también un `Module`: una red se usa igual que una capa. Esta es la idea central de PyTorch, y la verás otra vez en la semana 6 con el perceptrón multicapa (MLP, *multilayer perceptron*).

```python
class Sequential(Module):
    def __init__(self, *layers):
        self.layers = list(layers)

    def forward(self, x):
        for layer in self.layers:
            x = layer(x)
        return x

    def parameters(self):
        params = []
        for layer in self.layers:
            params.extend(layer.parameters())
        return params

    def __repr__(self):
        return "Sequential(" + ", ".join(repr(layer) for layer in self.layers) + ")"


net = Sequential(Linear(3, 4, rng), ReLU(), Linear(4, 2, rng))
print(net)
print("salida:", net(x).shape)
```
