### Recordatorio con un ejemplo resuelto

Una **clase** (*class*) es un molde; cada objeto que creas con ella es una **instancia** (*instance*). Piensa en un molde de galletas: el molde es la clase, cada galleta es una instancia.

Mira esta clase `PiggyBank` (una alcancía). Léela línea por línea y luego ejecútala.

````python
class PiggyBank:
    def __init__(self):
        # __init__ se ejecuta solo al crear la alcancía.
        # self es "esta alcancía": cada alcancía guarda su propio dinero.
        self.coins = []

    def deposit(self, amount):
        # Un método es una función que vive dentro de la clase.
        # Siempre recibe self primero.
        self.coins.append(amount)

    def total(self):
        return sum(self.coins)


mine = PiggyBank()      # crear una instancia: Python llama a __init__
mine.deposit(5)
mine.deposit(10)
yours = PiggyBank()     # otra instancia, con sus propias monedas
yours.deposit(1)

print("mi alcancía:", mine.coins, "→ total", mine.total())
print("tu alcancía:", yours.coins, "→ total", yours.total())
````

## Ejercicio 2: métodos especiales

Los métodos con doble guion bajo (*dunder methods*) conectan tu clase con Python:

- `__len__` hace funcionar `len(obj)`.
- `__repr__` decide qué se muestra al imprimir el objeto. Sin él, Python muestra algo como `<__main__.LabelCounter object at 0x7f3a...>`, que no dice nada.

En vez de copiar y pegar, usamos **herencia** (*inheritance*): `class B(A)` crea una clase `B` que recibe todo lo que tiene `A` y le agrega cosas. Así, `PrettyLabelCounter` ya tiene `__init__`, `add` y `most_common` sin escribirlos de nuevo.

Ejemplo resuelto, con `PiggyBank`:

````python
class PrettyPiggyBank(PiggyBank):   # hereda __init__, deposit y total
    def __len__(self):
        return len(self.coins)      # cuántas monedas hay

    def __repr__(self):
        return f"PrettyPiggyBank({len(self)} monedas, total={self.total()})"


bank = PrettyPiggyBank()
bank.deposit(5)
bank.deposit(10)
print(len(bank))
print(bank)
````
